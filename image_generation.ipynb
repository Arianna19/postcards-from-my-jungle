```python
import torch
from diffusers import StableDiffusionPipeline
from PIL import Image
import os
from tqdm import tqdm
```

```python
model_id = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
)

pipe = pipe.to("cuda")
pipe.enable_attention_slicing()
```

```python
os.makedirs("images", exist_ok=True)
```

```python
prompts = [
    "lo-fi surreal jungle creature, hybrid animal, painterly, soft diffusion, abstract",
    "dreamlike rainforest beast, strange fur patterns, glowing eyes, low resolution aesthetic",
    "imagined jungle mammal, blurry texture, oil painting style",
    "fantasy tropical animal, distorted proportions, muted colors",
]

num_images = 50

for i in tqdm(range(num_images)):
    prompt = prompts[i % len(prompts)]

    image = pipe(
        prompt,
        height=512,
        width=512,
        guidance_scale=7.5,
        num_inference_steps=30
    ).images[0]

    image.save(f"images/jungle_beast_{i+1}.png")
```
