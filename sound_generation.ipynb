```python
import torch
import os
import scipy.io.wavfile as wavfile
from diffusers import AudioLDMPipeline
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)
```

```python
pipe = AudioLDMPipeline.from_pretrained(
    "cvssp/audioldm",
    torch_dtype=torch.float16 if device == "cuda" else torch.float32
)

pipe = pipe.to(device)
```

```python
os.makedirs("sounds", exist_ok=True)
```

```python
sound_prompts = [
    "A gigantic jungle monkey monster screeching with metallic echoes, chaotic animal noises",
    "A tiny flying jungle beast with high-pitched screeches and erratic flapping wings",
    "A deep forest creature making distorted roars and echoing gurgles",
    "Invisible nocturnal jungle animals with eerie, glitchy calls and rustling sounds"
]

durations = [5, 3, 6, 4]  # seconds

for i, prompt in enumerate(sound_prompts):
    audio = pipe(
        prompt,
        num_inference_steps=50,
        audio_length_in_s=durations[i]
    ).audios[0]

    wavfile.write(
        f"sounds/jungle_sound_{i+1}.wav",
        rate=16000,
        data=audio
    )

    print(f"Saved jungle_sound_{i+1}.wav")
```
